```python
analyticsamplefiltered['weighted_public_var']
```

```python
workingsample = analyticsamplefiltered
workingsample.head()
```

Up to this point, I have now completed through Step 5, and created a weighted public variable. This is an additional column appended to the dataframe that includes the weight in time period t_1 multiplied by the value in the joined_state_local value in that same row (which is a 1 or a 0 in every row, depending on whether that household joined the public sector workforce that time period or not), for each and every row of the dataframe. The values in this column in each individual row are on their own useless to us at this time, but when the column is summed, should give us the weighted value of the population that joined the public sector workforce in this period. Please also note that I am now working with the same data, but renamed it workingsample for ease and to make sure that any changes made to this new dataframe do not impact or interfere with the original data I am working with.

```python
illinois_df = workingsample[workingsample['state_t'] == 'illinois']
illinois_df.head(5)
```

```python
illinois_total_weight = illinois_df['weight_t1'].sum()
print (illinois_total_weight)

illinois_total_weighted_public = illinois_df['weighted_public_var'].sum()
print (illinois_total_weighted_public)

illinois_recruitment_rate = illinois_total_weighted_public/illinois_total_weight
print (illinois_recruitment_rate)
```
