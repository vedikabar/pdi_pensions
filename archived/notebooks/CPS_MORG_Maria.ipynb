```python
# creating dfs with only Illinois values
tab208 = morg08[morg08['state'] == 33]
tab209 = morg09[morg09['state'] == 33]
tab210 = morg10[morg10['state'] == 33]
tab211 = morg11[morg11['state'] == 33]
tab212 = morg12[morg12['state'] == 33]
tab213 = morg13[morg13['state'] == 33]
tab214 = morg14[morg14['state'] == 33]
# 2015 doesn't have a column titled 'state', not sure why
```

```python
all_IL = pd.concat([tab208, tab209, tab210, tab211, tab212, tab213, tab214], ignore_index=True)
all_IL = all_IL[all_IL['class94'] == 2] # Public sector only
all_IL = all_IL[['year', 'earnhre', 'age', 'sex']]
all_IL
```

```python
# create separate dataframes for 2008 - 2010 and 2012 - 2014
pre_2011 = all_IL[(all_IL['year'] >= 2008) & (all_IL['year'] <= 2010)]
post_2011 = all_IL[(all_IL['year'] >= 2012) & (all_IL['year'] <= 2014)]
```

```python
results = []

# stats for hourly wage
pre_wage_mean = (pre_2011['earnhre']/100).mean()
post_wage_mean = (post_2011['earnhre']/100).mean()
wage_diff = post_wage_mean - pre_wage_mean
wt_stat, wp_val = stats.ttest_ind((pre_2011['earnhre']/100).dropna(), (post_2011['earnhre']/100).dropna())
results.append(['Hourly Wage', pre_wage_mean, post_wage_mean, wage_diff, wp_val])

# stats for age
pre_age_mean = pre_2011['age'].mean()
post_age_mean = post_2011['age'].mean()
age_diff = post_age_mean - pre_age_mean
at_stat, ap_val = stats.ttest_ind(pre_2011['age'].dropna(), post_2011['age'].dropna())
results.append(['Age', pre_age_mean, post_age_mean, age_diff, ap_val])
results

# stats for share
```
